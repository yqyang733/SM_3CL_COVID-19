```python
import os
from pymol import cmd
```

```python
with open("RCSBID.txt") as f:
    f1 = f.readlines()
pdbs = f1[0].strip().split(",")
pdbs
```

```python
for i in pdbs:
    cmd.load(os.path.join(i, i+".pdb.gz"), i)
    cmd.remove('solvent')
    cmd.remove('inorganic')
    cmd.remove('resn DMS')
    cmd.create("PRO", "chain A and polymer.protein")
    cmd.create("LIG", "chain A and organic")
    cmd.h_add("LIG")
    cmd.alter("LIG","resn='LIG'")
    cmd.save(os.path.join(i, i+".pdb"), "PRO")
    cmd.save(os.path.join(i, i+".mol2"), "LIG")
    cmd.delete("all")
```
