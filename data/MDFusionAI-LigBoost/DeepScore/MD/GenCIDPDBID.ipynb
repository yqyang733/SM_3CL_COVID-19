```python
print("aaaa")
dict_smipdbid = dict()
with open("SMILESPDBID.csv") as f:
    f1 = f.readlines()
for i in f1:
    line = i.strip().split(",")
    dict_smipdbid[line[0]] = line[1]
print(dict_smipdbid)

dict_smiactivity = dict()
with open("SMILES_activity.csv") as f:
    f1 = f.readlines()
for i in f1:
    line = i.strip().split(",")
    dict_smiactivity[line[0]] = [line[1],line[2]]
print(dict_smiactivity)

rt = open("CIDActivityPDB.csv", "w")
print(dict_smiactivity.keys())
for i in dict_smiactivity.keys():
    print(i)
    if i in dict_smipdbid:
        rt.write(dict_smiactivity[i][0]+","+i+","+dict_smiactivity[i][1]+","+dict_smipdbid[i]+"\n")
    else:
        rt.write(dict_smiactivity[i][0]+","+i+","+dict_smiactivity[i][1]+",Dock\n")
rt.close()
```
